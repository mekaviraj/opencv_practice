```python
import cv2
import numpy as     np
import matplotlib.pyplot as plt
```

```python
blank = np.zeros((512, 512,3), np.uint8)
plt.imshow(blank)
# purpple = np.zeros((512, 512), np.uint8)
# plt.imshow(purpple) 
```

```python
blank[:] = 255, 0, 255
plt.imshow(blank)
```

```python
blank[:, :] = 255, 0, 0
plt.imshow(blank)
```

```python
#RECTANGLE

blank[200:300, 100:400] = 255, 0, 255
plt.imshow(blank)
```

```python
#square
blank[100:400, 100:400] = 0, 255, 0
plt.imshow(blank)
```

```python
#drawing a proper rectangle
rect = cv2.rectangle(blank, (100, 100), (400, 400), (0, 255, 0), thickness=2)
rect2 = cv2.rectangle( blank, (150, 150), (350, 350), (255, 0, 0), thickness=2)
plt.imshow(rect)

# image, top-left corner, bottom-right corner, color (BGR), thickness
```

```python
blank =  np.zeros((512, 512, 3), np.uint8)
plt.imshow(blank)
```

```python
cir = cv2.circle(blank, (256, 256), 100, (0, 50, 255), thickness=10) 
#img , center, radius, color (BGR), thickness
plt.imshow(cir)
```

```python
line = cv2.line(blank, (0, 0), (511, 511), (255, 255, 255), thickness=5)
plt.imshow(line)
#img, start point, end point, color (BGR), thickness
text = cv2.putText(blank, "OpenCV", (200, 200), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), thickness=2)
#img, text, org, font, fontScale, color (BGR), thickness
plt.imshow(text)
```
