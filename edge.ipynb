```python
vid = cv.VideoCapture(0)

while vid.isOpened():
    ret, frame = vid.read()
    if not ret:
        print("Video has ended or cannot be read.")
        break

    cv.imshow("Frame", frame)
    if cv.waitKey(1) & 0xFF == ord('q'):
        print("Exiting...")
        break

vid.release()
cv.destroyAllWindows()
```

```python
#erosion and dilation 
import numpy as np
import cv2 as cv
kernel = np.ones((1,2) , np.uint8)
img = cv.imread("C:\\Users\\VIRAJ M\\Desktop\\opencv\\img2.jpg")
img = cv.resize(img, (400, 400))
grey = cv.cvtColor(img, cv.COLOR_BGR2GRAY) 
canny = cv.Canny(grey, 100, 200) 
erode_img = cv.erode(canny, kernel, iterations=1) 
```

```python
cv.imshow("Eroded Image", erode_img)
dilate_img = cv.dilate(erode_img, kernel, iterations=1)
cv.imshow("Dilated Image", dilate_img)
cv.waitKey(0)   
cv.destroyAllWindows()
```
