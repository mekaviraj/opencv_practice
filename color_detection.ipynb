```python
#color detection
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
vid = cv.VideoCapture(0)
while True:
    ret, frame = vid.read()
    if not ret:
        break
    hsv = cv.cvtColor(frame, cv.COLOR_BGR2HSV)
    # hue saturation value
    # HSV values for red color (red is at the ends of the hue spectrum)
    red_lower1 = np.array([0, 120, 70], np.uint8)
    red_upper1 = np.array([10, 255, 255], np.uint8)
    red_lower2 = np.array([170, 120, 70], np.uint8)
    red_upper2 = np.array([180, 255, 255], np.uint8)
    # Combine two masks for red color
    red_mask1 = cv.inRange(hsv, red_lower1, red_upper1)
    red_mask2 = cv.inRange(hsv, red_lower2, red_upper2)
    red_mask = cv.add(red_mask1, red_mask2)
    kernel = np.ones((5, 5), np.uint8)  # for dilation
    red_mask = cv.dilate(red_mask, kernel)
    res = cv.bitwise_and(frame, frame, mask=red_mask)
    contours, _ = cv.findContours(red_mask, cv.RETR_TREE, cv.CHAIN_APPROX_SIMPLE)

    # for creating a bounding box around the detected color
    for cnt in contours:
        area = cv.contourArea(cnt)
        if area > 500:
            x, y, w, h = cv.boundingRect(cnt)
            frame = cv.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), thickness=2)
            cv.putText(frame, "Red", (x, y - 10), cv.FONT_HERSHEY_SIMPLEX, 1.0, (0, 255, 0), 2)

    cv.imshow("Frame", frame)
    if cv.waitKey(1) & 0xFF == ord('q'):
        break
vid.release()
cv.destroyAllWindows()
```
